## 2. Categories and products with highest and lowest revenue

```python
categories = df["Category"].unique()

for category in categories:
    category_data = df[df["Category"] == category]
    category_revenue = category_data["Revenue"].sum()
    share = category_revenue / df["Revenue"].sum() * 100
    print(category, "Revenue =", category_revenue, "Share =", round(share, 2), "%")

products = df["Product"].unique()

for product in products:
    product_data = df[df["Product"] == product]
    print(product, "Revenue =", product_data["Revenue"].sum())

print("Highest revenue transactions:")
print(df.nlargest(3, "Revenue")[["Product", "Revenue"]])

print("Lowest revenue transactions:")
print(df.nsmallest(3, "Revenue")[["Product", "Revenue"]])
```

## 3. Branch performance

```python
branches = df["Branch"].unique()

for branch in branches:
    branch_data = df[df["Branch"] == branch]
    print(branch, "Revenue =", branch_data["Revenue"].sum())
```

## 4. Sales by month and weekday

```python
print("Month wise sales:")
for month in months:
    month_data = df[df["Month"] == month]
    print(month, "=", month_data["Revenue"].sum())

print("Weekday wise sales:")
weekdays = df["Weekday"].unique()

for day in weekdays:
    day_data = df[df["Weekday"] == day]
    print(day, "=", day_data["Revenue"].sum())
```

## 5. Average transaction value by category and branch

```python
print("Average revenue by category:")
for category in categories:
    category_data = df[df["Category"] == category]
    print(category, "=", category_data["Revenue"].mean())

print("Average revenue by branch:")
for branch in branches:
    branch_data = df[df["Branch"] == branch]
    print(branch, "=", branch_data["Revenue"].mean())
```

## 6. Discounts, quantity and transaction value

```python
discounts = df["Discount"].unique()

for discount in discounts:
    discount_data = df[df["Discount"] == discount]
    print("Discount =", discount, "%")
    print("Average quantity =", discount_data["Quantity"].mean())
    print("Average revenue =", discount_data["Revenue"].mean())
```

## 7. Customers who make frequent or high-value purchases
